````python
import httpx
import asyncio

async def fetch_profile():
    # 1. Use AsyncClient instead of Client
    async with httpx.AsyncClient(base_url="http://127.0.0.1:8080") as client:
        try:
            # 2. Add 'await' before the get request
            r = await client.get('/profile')
            
            print(r)
            print(r.url)
            print(r.status_code)
            print(r.headers)
            print(r.headers['content-type'])
            print(r.text)
            print(r.encoding)

            print(client.cookies)
            print(client.headers)
            print(client.base_url)
            print(client.timeout)
            print(client.auth)

        except httpx.HTTPError as e:
            print(f"Error fetching profile: {e}")

# 3. How to run the async function
# If you are running this in a Jupyter Notebook, just run:
await fetch_profile()

# If you are running this in a standard python file, use this instead:
# asyncio.run(fetch_profile())
````

# data packages transmission between client and server
When a client (like httpx) and a server (like FastAPI/Uvicorn) talk to each other, they send data packages back and forth. This exchange is split into two independent parts: the Request (sent by the client) and the Response (sent by the server). [1, 2, 3, 4, 5] 
Each package is like a physical mail delivery. It contains a label on the outside (Headers) and the actual items inside (Content/Body). [6, 7, 8]

 
------------------------------
## 1. The HTTP Request (From Client to Server) [9] 
   The request is the package the client builds and sends over the internet to ask the server to do something. [10, 11, 12] 

## Key Parts:

* The Request Line: The very top line. It has the action method (like POST or GET) and the URL path (like /login). [13, 14, 15, 16, 17] 
* Request Headers (The Mailing Label): Information generated by the client before sending. It includes:
* host: Where the package is going.
   * user-agent: The software making the call (e.g., httpx).
   * content-type: Tells the server how to read the payload if data is included. [18, 19, 20, 21, 22] 
* Request Body / Payload (The Box Content): The actual data being sent, like {"username": "Alice"}. It is kept completely separate from the headers. [23, 24, 25, 26] 

## Importance:

   It tells the server exactly who is calling, what action they want to take, and provides the raw data needed to complete the task. [27, 28]

------------------------------
## 2. The HTTP Response (From Server to Client)
The response is the package the server creates and sends back after it processes the client's request. [29, 30, 31] 
## Key Parts:

* Status Code: A three-digit number showing if the job succeeded (like 200 OK) or failed (like 404 Not Found). [32, 33, 34, 35, 36] 
* Response Headers (The Return Label): Extra information added by the server. It includes:
* date: The exact time the server processed the job.
   * server: The name of the backend technology (e.g., uvicorn).
   * set-cookie: A critical security header that tells the client to save session details. [37, 38, 39, 40, 41] 
* Response Body / Content: The server's final reply message, usually in JSON format (like {"message": "Logged in successfully!"}). [42, 43] 

## Importance:

It delivers the results of the job back to the user and gives the client instructions on what to do next (like storing a cookie for logging in). [44] 

------------------------------
## 📝 Important Notes to Remember

* No Mixing: The server cannot modify or inject code into the client's request headers. It can only read them and build its own separate response headers. [45, 46, 47, 48] 
* Viewing via Python: In httpx, you can inspect the client's outgoing labels using r.request.headers, and the server's incoming labels using r.headers.
* Payload Location: Data sent via json= or data= parameters never sits inside the headers. It always travels inside the Body section at the bottom of the message. [49, 50]

# Seession in Server Client interaction

Think of a session as a private, ongoing conversation between a specific client (like your browser or a Python script) and a server.
Instead of treating every single request as a total stranger, the server uses a session to say, "Ah, it’s you again! I remember what we were talking about."

The Analogy: A Coffee Shop Tab

Without a Session (Stateless): Every time you walk up to the counter, you have to tell the barista your name, give them your credit card, and explain that you like oat milk.

With a Session: You walk in, open a "tab" (the session). The barista hands you a wristband (the Session ID). For the rest of the afternoon, you just show the wristband to get drinks, and they add it to your profile. When you leave, the tab is closed.

httpx, which is the modern standard for asynchronous HTTP requests in Python (very similar to aiohttp or requests).

````python
import asyncio
import httpx

async def main():
    # 1. Create a persistent session (Client)
    async with httpx.AsyncClient(base_url="http://127.0.0.1:8080") as session:
        
        # 2. Log in to create the session on the server
        print("Logging in...")
        login_response = await session.post("/login", params={"username": "Alice"})
        print("Server says:", login_response.json())
        
        print("\n--- Doing other things... ---\n")
        
        # 3. Visit the profile page (The session automatically sends the cookie!)
        print("Accessing profile...")
        profile_response = await session.get("/profile")
        print("Server says:", profile_response.json())

        return session

session = await main()
````

# cookies Basics

Let's clear up how the server tracks this, because this is where the real magic of a **session** happens.

You are entirely right: the cookie itself lives in the client's RAM (or hard drive). **The server does not store your cookie.** Instead, the server stores a **Session Database** (or a lookup table in its own RAM). The cookie is simply the *key* used to look up that data.

---

### The Parallel RAM Analogy: The Coat Check

Think of going to a restaurant with a coat check:

1. **The Server's RAM (The Coat Room):** The restaurant has a room full of coats (user data, shopping carts, profile info).
2. **The Cookie in your RAM (The Plastic Ticket):** The attendant hands you a plastic ticket with the number `#442`. You put that ticket in your pocket (Client RAM).

The restaurant doesn't keep your ticket, and you don't keep their coats. But when you show them ticket `#442`, they look at their racks, match the number, and hand you your specific coat.

---

### Step-by-Step: How Both RAMs Interact

Here is exactly what happens in the computer memory of both machines during our FastAPI example.

```
CLIENT RAM (HTTPX Client)              SERVER RAM (FastAPI Backend)
┌───────────────────────┐              ┌──────────────────────────┐
│                       │  1. Login    │                          │
│  "I am Alice"         │ ───────────> │ Validates Alice.         │
│                       │              │ Generates Token: "XYZ7"  │
│                       │  2. Cookie   │ Saves in RAM dict:       │
│ Stores Cookie:        │ <──────────_ │ {"XYZ7": "Alice's Data"} │
│  { session_id: XYZ7 } │              │                          │
└───────────────────────┘              └──────────────────────────┘

```

#### Step 1: The Server Allocates RAM (The Login)

When you call `/login?username=Alice`, the FastAPI server runs code to create a session.

* **On the Server:** It creates a dictionary in its RAM that looks like this:
```python
# Server Memory
ACTIVE_SESSIONS = {
    "XYZ79934": {"username": "Alice", "role": "admin", "login_time": "10:00 AM"}
}

```


* **The Handoff:** The server finishes the request by telling the client: *"Hey, write down the code `XYZ79934` on a sticky note (cookie)."*

#### Step 2: The Client Holds the Key

Your `httpx.AsyncClient` receives that string and saves it in its own RAM (`client.cookies`).

* The client **does not** know Alice's role is admin, and it **does not** know her login time. It only knows the random string `XYZ79934`.

#### Step 3: The Matchmaking (The Profile Request)

When you call `client.get("/profile")`, the client automatically attaches that string to the request headers.

When FastAPI receives the request, it does a RAM lookup:

```python
# What FastAPI does under the hood
incoming_cookie = request.headers.get("Cookie")  # Gets "XYZ79934"

# It checks its own RAM dictionary
if incoming_cookie in ACTIVE_SESSIONS:
    user_data = ACTIVE_SESSIONS[incoming_cookie]
    return f"Welcome back, {user_data['username']}!"

```

---

### Where does the Server store this dictionary?

In our tiny FastAPI code example, we didn't explicitly write a database, so FastAPI was just keeping track of it implicitly. In production web development, the server saves that session lookup table in one of two places:

1. **In-Memory Database (Redis):** A lightning-fast database that lives entirely in the server's RAM specifically designed to match Session IDs to User Data instantly.
2. **A SQL Database (PostgreSQL/MySQL):** Saved onto the server's hard drive so that even if the server crashes or restarts, nobody gets logged out.

### Summary

The server doesn't store the cookie; **it stores the data *associated* with the cookie.** The cookie is just a pointer, a unique ID card. When the client passes that ID card over the network, the server uses it to query its own internal RAM memory and find out who you are.


# The 5 Main Categories of HTTP Status Codes
HTTP status codes are three-digit numbers. The first digit tells you the general result of your request. [5, 6] 

* 
* 1xx (Informational): The server received the request and is still processing it.
* 2xx (Success): Everything went well. The request was received and accepted.
* 3xx (Redirection): The requested page moved. You need to look in a new place.
* 4xx (Client Error): Something went wrong on your end. The page might not exist, or you might need a password.
* 5xx (Server Error): The server made a mistake or crashed while trying to load the page. [7, 8, 9, 10, 11] 
*

## How to Get the Status Code in Python httpx
To get the status code using the Python httpx library, you make a request and look at response.status_code. [17, 18] 
Here is a short, simple code example:

## 2. Print the status code number

## 3. Print the text name of the status code
print("Status Phrase:", response.phrase)
## 4. Check if the request was successful (2xx codes)if response.is_success:
    print("The request worked!")

## Helpful Shortcuts in httpx [19] 
Instead of remembering all the numbers, you can use these built-in true/false shortcuts:

* 
* response.is_informational (True for 1xx codes)
* response.is_success (True for 2xx codes)
* response.is_redirect (True for 3xx codes)
* response.is_client_error (True for 4xx codes)
* response.is_server_error (True for 5xx codes)
* response.is_error (True for both 4xx and 5xx codes) [20, 21, 22, 23, 24]
