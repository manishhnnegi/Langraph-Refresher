## Initialising and Invoking a model

- The foundation of an agent is the model inside it.

```python
from langchain.chat_models import init_chat_model # initialise the model using the init_chat_model function

model = init_chat_model(model="gpt-5-nano")
```

```python
model
```

```python
response = model.invoke("What is the capital of France?") # invoke the model with the invoke method
response # Get AIMessage object which has content attribute as the response. There is metadata which has token usage
```

```python
print(response.content)
```

```python
from pprint import pprint
pprint(response.response_metadata)
```

## Customising Our Model

```python
model = init_chat_model(
    model="gpt-5-nano",
    # Kwargs passed to the model
    temperature=1.0 # controls the randomness of the output. Higher temperature means more creative output, lower temperature means more deterministic output
)
# max tokens controls how long the response can be and 
# timeout controls max time to wait for a response from the model
# model_kwargs is a dictionary of keyword arguments to pass to the model

response = model.invoke("What's the capital of France?")
pprint(response.content)
```

```python
response = model.invoke("Tell me lunar capital from Marvel or Star Wars?")
pprint(response.content)
```

## Initialising and Invoking an Agent

```python
from langchain.agents import create_agent 
# create_agent built on top of Lang graph

agent = create_agent(model=model) # create an agent with the model. in this case the model is gpt-5-nano
```

```python
agent = create_agent(model="gemini-3-flash-preview",model_provider="google_genai")
```

```python
agent = create_agent("gpt-5-nano")
```

```python
from langchain.messages import HumanMessage
# to create an agent we just invoke it, we pass in a dictionary with a messages key using the HumanMessage function to specify it is coming from the user
response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="Who won the 2011 Cricket World Cup?")
        ]
    }
)
```

```python
from pprint import pprint

pprint(response)
```

```python
print(response['messages'][-1].content)
```

```python
from langchain.messages import AIMessage

# As we are passing in a dictionary, it can also contain a chat history with several HumanMessage and AIMessage.

response = agent.invoke(
    {
        "messages": [
            HumanMessage(content="Who won the 2010 Football World cup?"),
            AIMessage(content="The 2010 World Cup was won by Spain. They beat the Netherlands 1-0 in the final."),
            HumanMessage(content="Who was the top scorer in the tournament?")
        ]
    }
)

pprint(response)
```

## Streaming Output

- In AI systems, we might be facing the issue of latency.
- Some software systems may measure response time in milliseconds.
- We will be running agent with response times of seconds, even minutes already as we handed it a few messages.
- A way to reduce the **perceived latency** of the system is by streaming tokens to the user as they appear.

```python
for token, metadata in agent.stream(
    {"messages": [HumanMessage(content="Who has scored the most goals in the UEFA Champions League?")]},
    stream_mode="messages"
):
    # token is a message chunk with token content
    # metadata contains which node produced the token
    if token.content: # Check if there is actual content
        print(token.content, end="", flush=True) # Print token
```
