# Memory

- A basic feature that we expect from any chat bot is its ability to maintain memory over the length of a conversation.
- Agents we have been creating so far do not have that ability.

```python
from dotenv import load_dotenv

load_dotenv()
```

# No Memory

```python
from langchain.agents import create_agent

agent = create_agent(
    "gpt-5-nano"
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Cosmic and my favorite color is white.")

response = agent.invoke(
    {"messages": [question]}
)
```

```python
response['messages'][1]
```

```python
from pprint import pprint

pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]}
)

pprint(response)
```

# Memory

- In langchain, we have something called state, which is the memory of our agent.
- State is not saved from one run to another. Due to this agent's memory is being wiped. We have to save the state so that agent remembers the previous messages.
- This is done using checkpointer

![](https://i.imgur.com/o36bAGu.png)

```python
from langgraph.checkpoint.memory import InMemorySaver

agent = create_agent(
    "gpt-5-nano",
    checkpointer=InMemorySaver()
)
```

```python
from langchain.messages import HumanMessage

question = HumanMessage(content="Hello my name is Cosmic and my favourite color is blue")
config = {"configurable": {"thread_id": "1"}}

response = agent.invoke(
    {"messages": [question]},
    config,
)

pprint(response)
```

```python
question = HumanMessage(content="What's my favourite colour?")

response = agent.invoke(
    {"messages": [question]},
    config
)

pprint(response)
```
