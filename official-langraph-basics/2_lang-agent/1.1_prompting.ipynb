# Prompting

- The easiest way to customize the performance of a chat model is with system prompts.
- Example below shows when we ask the model for the capital of moon, and in this case it will give us the information because we have given the system prompt that we are in a fictional world.

```python
from langchain.agents import create_agent
from langchain.messages import HumanMessage

agent = create_agent("gpt-5-nano")

question = HumanMessage(content="What is the capital of the moon?")

response = agent.invoke(
    {
        "messages": [question]
    }
)

print(response)
```

```python
print(response["messages"][-1].content)
```

```python
from pprint import pprint

pprint(response['messages'][-1].response_metadata)
```

```python
pprint(response['messages'][-1].usage_metadata)
```

# Including a system prompt

- Being able to engineer the perfect prompt is both an art and a science.

```python
system_prompt = "You are a science fiction writer. Create a capital city at the users request."

scifi_agent = create_agent(
    model="gpt-5-nano", 
    system_prompt=system_prompt
)

response = scifi_agent.invoke(
    {
        "messages": [question]
    }
)
```

```python
print(response)
print(response["messages"][-1].content)
```

## Few Shot Examples

```python
system_prompt = """
You are a science fiction writer. Create a capital city at the users request.

User: What is the capital of mars?
Scifi Writer: New Olympus.

User: What is the capital of Venus?
Scifi Writer: Nova Roma.
"""

scifi_agent = create_agent(
    model="gpt-5-nano",
    system_prompt=system_prompt
)

response = scifi_agent.invoke(
    {
        "messages": [question]
    }
)
```

```python
print(response)
print(f"Response from Scifi Writer: {response['messages'][-1].content}")
```

# Structured Prompt

- Another technique of prompting is to provider the structure we want our prompt to use within the prompt itself.

```python
system_prompt = """
You are a science fiction writer. Create a capital city at the users request.

Please keep the below structure:

Name: The name of the city

Location: Where is it based

Vibe: 2-3 words to describe the vibe of the city

Economy: Main Industry

"""

scifi_agent = create_agent(
    model="gpt-5-nano",
    system_prompt=system_prompt
)

response = scifi_agent.invoke(
    {
        "messages": [question]
    }
)
```

```python
response
```

```python
response['messages'][1].content
```

# Structured Prompt

- Instead of providing the structure to the model within the prompt, what we do is we provide an output schema that we want model to fill out.
- This is super helpful if we want to use the output in the functions downstream.
- We can call fields within the output instead of having a block of text which we cannot really get values out of. 
- To do this we leverage pydantic.

```python
from pydantic import BaseModel

class CapitalInfo(BaseModel):
    name: str
    location: str
    vibe: str
    economy: str
    
agent = create_agent(
    model="gpt-5-nano",
    system_prompt="You are a science fiction writer. Create a capital city at the users request.",
    response_format=CapitalInfo
)

response = agent.invoke({
    "messages": [question]
})
```

```python
response
```

```python
print(response["structured_response"].name)
print(response["structured_response"].location)
print(response["structured_response"].vibe)
print(response["structured_response"].economy)
```
